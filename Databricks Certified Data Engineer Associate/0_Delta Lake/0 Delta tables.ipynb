```sql
%sql
use catalog workspace;
use schema default
```

```sql
%python
spark.conf.set("spark.databricks.delta.optimizeWrite.enabled", "false")
spark.conf.set("spark.databricks.delta.autoCompact.enabled", "false")
```

```sql
create table employees(
  id int,
  name string,
  salary double
)
using delta
```

```sql
ALTER TABLE employees DISABLE PREDICTIVE OPTIMIZATION;
ALTER TABLE employees SET TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'false',
  'delta.autoOptimize.autoCompact' = 'false',
  'delta.enableDeletionVectors' = 'false'
);
```

```sql
insert into employees values (1, 'John', 200), (2, 'Mary', 700), (3, 'Mike', 900);
insert into employees values (4, 'Ivan', 400), (5, 'Anna', 800), (6, 'Roman', 600);
insert into employees values (7, 'Andriy', 300);
select * from employees;
```
